### 07) WAP to print sum of series 1 – 2 + 3 – 4 + 5 – 6 + 7 ... n.

```python
sum = 0
n = int(input("Enter value of n :       "))

for i in range(1,n+1):
    if(i%2==0):
        sum-=i
    if(i%2!=0):
        sum+=i
print(sum)
```

### 12) WAP to find whether the given number is Prime or not.

```python
import math
num1 = int(input("Enter a number :          "))

if num1<=1:
    print(f"not prime")
else:
    for i in range(2, int(math.isqrt(num1))+1):
        if(num1%i==0):
            print("not prime")
            break
    else:
        print(f"{num1} is prime")
```

### 13) WAP to print sum of digits of given number.

```python
num1 = int(input("Enter number :        "))
temp = abs(num1)
sum = 0

if(temp==0):
    sum = 0
else:
    while(temp>0):
        digit = temp%10
        sum += digit
        temp = temp//10
    print(sum)

```

### 14) WAP to check whether the given number is Palindrome or not.

```python
num1 = int(input("Enter number :        "))
reverse = 0
temp = abs(num1)

while(temp>0):
    digit = temp%10
    reverse = reverse*10+digit
    temp = temp//10
print(f"{reverse==num1}")

```

### 15) WAP to check whether the given number is an Armstrong Number or not.

```python
num = int(input("Enter a number :       "))

if num < 0:
    print(False)
elif num == 0:
    print(True)
else:
    total_digits = 0
    temp = num
    while temp > 0:
        total_digits += 1
        temp //= 10

    
    temp = num
    arm = 0
    while temp > 0:
        digit = temp % 10
        arm += digit ** total_digits
        temp //= 10

    print(arm == num)
```
