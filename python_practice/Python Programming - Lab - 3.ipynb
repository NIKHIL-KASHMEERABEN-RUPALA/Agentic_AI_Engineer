### 05) WAP to check whether the given year is leap year or not.
#### [If a year can be divisible by 4 but not divisible by 100 then it is leap year but if it is divisible by 400 then it is leap year]

```python
year = int(input("Enter the year :      "))

isLeapYear = True if (year%4==0 and year%100!=0 ) or (year%400==0) else False

if isLeapYear:
    print(f"This is a leap year")
else:
    print(f"This is not a leap year")
```

### 06) WAP in python to display the name of the day according to the number given by the user.

```python
num = int(input("Enter a number for the day :       "))
match(num):
    case 1:
        print(f"Monday")
    case 2:
        print(f"Tuesday")
    case 3:
        print(f"Wednesday")
    case 4:
        print(f"Thursday")
    case 5:
        print(f"Friday")
    case 6:
        print(f"Saturday")
    case 7:
        print(f"Sunday")
    case _:
        print(f"Invalid number entered")
        
```

### 07) WAP to implement simple calculator which performs (add,sub,mul,div) of two no. based on user input.

### 08) WAP to read marks of five subjects. Calculate percentage and print class accordingly. 
#### Fail below 35 
#### Pass Class between 35 to 45 
#### Second Class
#### between 45 to 60
#### First Class between 60 to 70
#### Distinction if more than 70

```python
sub1 = int(input("Enter marks of subject 1 :      "))
sub2 = int(input("Enter marks of subject 2 :        "))
sub3 = int(input("Enter marks of subject 3 :        "))
sub4 = int(input("Enter marks of subject 4 :        "))
sub5 = int(input("Enter marks of subject 5 :        "))

percentage = ((sub1+sub2+sub3+sub4+sub5)/5)*100

if percentage < 35:
    print(f"Failed ! ")
elif percentage >= 35 and percentage < 45:
    print(f"Pass class")
elif percentage >=45 and percentage <60:
    print(f"Second class")
elif percentage >=60 and percentage < 70:
    print(f"First class")
elif percentage>=70:
    print(f"Distinction")
    
```

### 10) WAP to calculate electricity bill based on following criteria. Which takes the unit from the user.
